### Loading the Dataset

In this section, the Palmer Penguins dataset is loaded directly from an online source using its URL. The dataset is stored in a pandas DataFrame, which allows for easy data manipulation and analysis.

The first five rows of the dataset are displayed to get an overview of its structure and the variables it contains.

```python
# URL of the penguins dataset
url = "https://raw.githubusercontent.com/mcnakhaee/palmerpenguins/refs/heads/master/palmerpenguins/data/penguins.csv"

# Read the dataset into a pandas DataFrame
penguins = pd.read_csv(url)

# Display the first 5 rows
penguins.head()
```

```python
penguins.describe()
```

## Problem 6: Visualizing Distributions

### Histogram - Body Mass (g)
[Matplotlib - Histograms](https://matplotlib.org/3.5.3/api/_as_gen/matplotlib.pyplot.hist.html)

```python
plt.hist(penguins['body_mass_g'].dropna())
plt.title("Histogram of Body Mass (g)")
plt.xlabel("Body Mass (g)")
plt.ylabel("Frequency")
```

### Box Plot - Body Mass (g)
[Matplotlib - Box Plots](https://matplotlib.org/3.5.3/api/_as_gen/matplotlib.pyplot.boxplot.html)

```python
plt.boxplot(penguins['body_mass_g'].dropna())
plt.title("Box Plot of Body Mass (g)")
plt.ylabel("Body Mass (g)")
```

### Comparison of Histogram and Box Plot
[The Power of Visualisation: Using Plots to Understand Data](https://medium.com/@wtc2189/the-power-of-visualization-using-simple-plots-to-understand-your-data-99bf1cbd10c1)
- The histogram shows how the values are distributed, including the shape of the data and how frequently values occur across different ranges.

- The box plot summarises the data using the median and quartiles, and it clearly highlights any outliers.

- While the histogram gives a more detailed view of the distribution, the box plot makes it easier to quickly see the centre of the data and identify extreme values.


## Problem 7: Investigating Relationships
[Matplotlib - Scatter Plots](https://matplotlib.org/3.5.3/api/_as_gen/matplotlib.pyplot.scatter.html)

```python
colours = {'Adelie': 'blue', 'Chinstrap': 'green', 'Gentoo': 'red'}
plt.scatter(penguins['body_mass_g'], penguins['bill_length_mm'], c=penguins['species'].map(colours))

plt.xlabel("Body Mass (g)")
plt.ylabel("Bill Length (mm)")
plt.title("Body Mass vs Bill Length")
```

### Relationship Between Body Mass and Bill Length
- The scatter plot shows a positive relationship between body mass and bill length, meaning that larger penguins tend to have longer bills.

- When the points are coloured by species, you can see three clear groups. Gentoo penguins are generally heavier, Adelie penguins are smaller, and Chinstrap penguins seem to be somewhere in the middle.

- This highlights that the relationship between body mass and bill length exists, and depends on the species of penguin.
